**Create a DataFrame containing the columns Name, Gender, and Age, and perform DataFrame slicing to: o Display the first two rows. o Display all rows from the third row onwards.**

```python
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Gender': ['F', 'M', 'M', 'M', 'F'],
    'Age': [25, 30, 35, 40, 45]
})

print("Original DataFrame:")
print(df)

print("\nFirst two rows:")
print(df.iloc[:2])

print("\nAll rows from the third row onwards:")
print(df.iloc[2:])
```

**Create a Pandas Series containing four random numbers, and display: o The complete Series. o The values stored in the Series. o The first two elements using head(). o The last three elements using tail().**

```python
df = pd.Series([3, 7, 8, 12])

print("Original Series:")
print(df)

print("\nValues stored in the Series:")
print(df.values)

print("\nFirst two elements using head():")
print(df.head(2))

print("\nLast three elements using tail():")
print(df.tail(3))
```

**Create a DataFrame containing the columns Name, Age, and Rating, and display the following DataFrame properties: o Transpose of the DataFrame. o Row and column labels using axes. o Data type of each column using dtypes. o Whether the DataFrame is empty using empty. o Number of dimensions using ndim. o Shape of the DataFrame using shape. o Total number of elements using size. o Values stored in the DataFrame using values. o Number of missing values in each column using isnull().sum().**

```python
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Age': [25, 30, 35, 40, 45],
    'Rating': [4.5, 3.8, 4.2, 3.9, 4.1]
})

print("Original DataFrame:")
print(df)

print("\nTranspose of the DataFrame:")
print(df.T)

print("\nRow and column labels using axes:")
print(df.axes)

print("\nData type of each column using dtypes:")
print(df.dtypes)

print("\nWhether the DataFrame is empty using empty:")
print(df.empty)

print("\nNumber of dimensions using ndim:")
print(df.ndim)

print("\nShape of the DataFrame using shape:")
print(df.shape)

print("\nTotal number of elements using size:")
print(df.size)

print("\nValues stored in the DataFrame using values:")
print(df.values)

print("\nNumber of missing values in each column using isnull().sum():")
print(df.isnull().sum())
```
