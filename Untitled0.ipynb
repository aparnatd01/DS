```python
import matplotlib.pyplot as pit
x=[1,2,3,4,5]
y=[10,20,30,40,50]
pit.plot(x,y)
pit.show()
```

```python
months=["jan","feb","mar","apr","may"]
sales=[100,150,120,180,200]
pit.plot(months,sales)
pit.title("monthly sales")
pit.xlabel("month")
pit.ylabel("sales")
pit.show()
```

```python
months=["jan","feb","mar","apr"]
sales1=[100,150,130,180]
sales2=[80,120,160,170]
pit.plot(months, sales1, label="product A")
pit.plot(months, sales2, label="product B")
pit.title("product sales comparison")
pit.xlabel("month")
pit.ylabel("sales")
pit.legend()
pit.show()
```

```python
students=["arun","binu","celine","david"]
marks=[75,85,65,90]
pit.bar(students,marks)
pit.title("student marks")
pit.xlabel("students")
pit.ylabel("marks")
pit.show()
```

```python
students=["arun","binu","celine","david"]
marks=[75,85,65,90]
pit.barh(students,marks)
pit.title("student marks")
pit.xlabel("students")
pit.ylabel("marks")
pit.show()
```

```python
import matplotlib.pyplot as plt
salary=[20,22,25,25,28,30,32,35,40,45]
pit.hist(salary, bins=5)
pit.title("employee salary distribution")
pit.xlabel("salary(₹ thousands)")
pit.ylabel("number of employees")
pit.show()
```

```python
hours=[1,2,3,4,5,6]
marks=[45,50,60,65,75,85]
pit.scatter(hours,marks)
pit.title("study hours vs marks")
pit.xlabel("study hours")
pit.ylabel("marks")
pit.show()
```

```python
subjects = ["python","java","dbms","ai"]
students= [30,25,20,25]
pit.pie(students, labels=subjects)
pit.title("student interest by subject")
pit.show()
```

```python
subjects = ["python","java","dbms","ai"]
students= [30,25,20,25]
pit.pie(students, labels=subjects,autopct='%1.1f%%')
pit.title("student interest by subject")
pit.show()
```

```python
import matplotlib.pyplot as plt
marks=[45,50,52,55,60,62,65,68,70,150]
pit.boxplot(marks)
pit.title("student marks -outlier detection")
pit.ylabel("marks")
pit.show()
```

```python
x=[1,2,3,4,5]
y1=[10,20,30,40,50]
y2=[50,40,30,20,10]
pit.subplot(1,2,1)
pit.plot(x,y1)
pit.title("increasing")
pit.subplot(1,2,2)
pit.plot(x,y2)
pit.title("decreasing")
pit.show()
```
